## Matplotlib
Matplotlib is a Python library used to

👉 create graphs and charts from data

👉 Helps you visualize data (understand it better)

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(10,20)
y = np.arange(20,30)
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Line Chart")
plt.plot(x,y)
plt.show()
```

```python
x = np.arange(0,10)
y = np.arange(20,30)
plt.scatter(x,y,c='g') # we can define the Dot Color
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Line Doted Chart")
plt.show()
```

```python
y=x*x
# plt.plot(x,y,'r') # (r) is color of line
# plt.plot(x,y,'r--') # (r--) this will print doted line -------- graph
plt.plot(x,y,'r*') # (r--) this will print star line * * * * graph
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Line Doted Chart")
plt.show()
```

```python
# Now Creating an Sub-plot
plt.title("Subplots")
plt.subplot(2,2,1)   #(2,2,1) means 2 Rows, 2 columns, position of the graph like 1,2,3,4 Index and (2,2) in first means we can only plot 4 sub plot 2 rows and 2 columns
plt.plot(x,y,'r--')
plt.subplot(2,2,2)
plt.plot(x,y,'g*--')
plt.subplot(2,2,3)
plt.plot(x,y,'bo')
plt.subplot(2,2,4)
plt.plot(x,y,'r*')
```

```python
# We can also Plot an Waves form
np.pi
```

```python
# We can also Plot an Waves form, Now Compute the x and y coordinates for points on a sine curve
x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)
plt.title("Sine wave form")

# Plot the points using matplotlib
plt.plot(x,y)
plt.show()
```

```python
# We can also Create an Bar Chart Using matplotlib
x = np.arange(0,7)
y = np.arange(10,17)

plt.bar(x,y,color='g')

plt.title("Bar Chart")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

# ✅ What is Histogram?

A Histogram is a graph that shows how data is distributed (spread).

👉 It groups numbers into ranges (bins) and shows how many values fall in each range.

✅ Why do we use it?

To understand data pattern

To see frequency (count)

To check data spread (low, medium, high values)

To detect skewness or outliers

```python
a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27])
plt.hist(a)
plt.title("Histogram")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

```python
# Box plot
data = [np.random.normal(0, std, 100) for std in range(1,4)]
#Rectangle box plot
plt.boxplot(data,vert=True,patch_artist=True);
```

```python
#Pie Chart Showing an Percentage
# Data to plot
labels = 'Java', 'Python', 'Rust', 'JavaScript'
sizes = [215, 130, 245, 210] # this are the percentage
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode = (0.2, 0, 0, 0) # We can give an space to the Sclie like don for Python

#plot
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True)
plt.axis('equal')
plt.show()
```

## Join Plot

A join plot allows to study the replationship between 2 numeric variables. The Central chart display their Correlation. It is usually a Scatterplot, a hexbin plot, a2D histogram & a 2D density plot

Univariate Analysis

 (Means Comarin two diffrent features with each other in current dataset is total_bill and tip)

```python
sns.jointplot(x='tip',y='total_bill', data=df,kind='hex')
```

```python
sns.jointplot(x='tip',y='total_bill', data=df,kind='reg')  # reg means regression
```
